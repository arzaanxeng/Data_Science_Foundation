```python
dataset = [
    101, 101, 101, 102, 105, 105, 106, 107, 108, 113, 114, 114, 115, 
    117, 127, 132, 134, 135, 137, 138, 140, 143, 147, 147, 147, 
    300, 340, 330  
]
```

```python
def trace_outliers(data) :
    thresh = 3
    m = np.mean(data)
    std = np.std(data)
    outliers = []
    
    for i in data:
        z_score = np.abs((i - m) / std) 
        if ( z_score > thresh ):
            outliers.append(i)
    return outliers
```

```python
trace_outliers(dataset)
```
