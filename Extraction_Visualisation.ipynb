```python
import pandas as pd
import matplotlib.pyplot as plt
import seaborn as sns
import warnings
warnings.filterwarnings('ignore')


df = pd.read_csv(r'IPL.csv')
```

```python
df.head()
```

```python

print(f"THE NUMBER OF ROWS ARE {df.shape[0]} AND COLUMNS ARE {df.shape[1]}.")
```

```python
matchwins = df['match_winner'].value_counts()
matchwins
```

```python

sns.barplot( x = matchwins.values , y = matchwins.index , palette = 'rainbow')
plt.title("TOP_TEAMS_OF_THE_TOURNAMENT")
```

```python
sns.countplot(x = df['toss_decision'] , color = 'gold' )
plt.title("TOSS_WINNING_RESULTS")
```

```python
df.head(2)
```

```python
mom = df['player_of_the_match'].value_counts().head(10)
mom
```

```python
sns.barplot( x = mom.values , y = mom.index , palette = "Spectral")
plt.title("TOP_PLAYERS_OF_THE_TOURNAMENT")
```

```python
count = df[df['toss_winner'] == df['match_winner']]['match_id'].count()
percentage = count*100/df.shape[0]
x = percentage.round(4)
x
```

```python
df.head(1)
```

```python
top_batters = df['top_scorer'].value_counts().head(10)
sns.barplot( x = top_batters.values , y = top_batters.index , palette = 'mako')
plt.title('TOP_SCORERS_PER_MATCH')
plt.xlabel('matches_played')
```

```python
bowl = df['best_bowling'].value_counts().head(10)
bowl
```

```python
sns.barplot( x = bowl.values , y = bowl.index , palette = 'coolwarm')
plt.title("TOP_BOWLERS_PER_MATCH")
```

```python
a = df.groupby('match_winner')['player_of_the_match'].count().sort_values(ascending = False)
a
```

```python
sns.barplot( x = a.values , y = a.index , palette = 'rainbow')
plt.title("Highest_Man_of_the_Matches_Award_Winning_Teams")
plt.xlabel("No_Of_Awards")
```
