```python
import numpy as np
import pandas as pd
import matplotlib.pyplot as plt
import seaborn as sns
```

```python
df = pd.read_csv(r'anime.csv')
```

```python
df.head()
```

```python
def extracting_name(text):
    data = ""
    check = False
    for i in range(len(text)):
        if text[i] == '(':
            return data
        else:
            data = data + text[i]
```

```python
df['Names'] = df['Title'].apply(extracting_name)
```

```python
df.head()
```

```python
def extracting_episodes(text):
    check = False
    data = ""
    for i in text:
        if i == ')':
            check = False
            return data
        if check == True:
            data = data + i
        if i == '(':
            check = True
```

```python
df['Episodes'] = df['Title'].apply(extracting_episodes)
```

```python
def extracting_timestamp(text):
    data = ""
    for i in range(len(text)):
        if text[i] == ')':
           for j in range(i+1 , i +20):
               data = data + text[j]

    return data
```

```python
df['Timestamp'] = df['Title'].apply(extracting_timestamp)
```
