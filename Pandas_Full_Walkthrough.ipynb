```python
import numpy as np
import pandas as pd
```

```python

data = {
    'Name': ['John', 'Anna', 'Peter', 'Linda'],
    'Age': [28, 34, 29, 42],
    'City': ['New York', 'Paris', 'Berlin', 'London'],
    'Salary': [65000, 70000, 62000, 85000]
}
df = pd.DataFrame(data)
```

```python
print(df)
```

```python
data_list = [
    ['John', 28, 'New York', 65000],
    ['Anna', 34, 'Paris', 70000],
    ['Peter', 29, 'Berlin', 62000],
    ['Linda', 42, 'London', 85000]
]
# You would have to craete a seperate list for defining the labels 
columns = ["Name" , "Age" , "City" , "Salary" ]
df = pd.DataFrame(data , columns = columns)
print(df)
```

**ADDING AND REMOVING A NEW COLUMN**

```python
df["Designation"] = ["Engineer" , "Doctor" , "Lawyer" , "Developer"]
```

```python
df
```

```python
df.drop("Designation" , axis = 1)
# To completely remove  the designation column use { inplace = True }
```

```python
df
```

**EXTRACTING DATA BASED UPON CONDITIONS**

```python
# Extract people from the dataset whose age is above 25
df[df["Age"] > 25  ]
```

```python
# Extract people from dataset who live in NewYork and London
df[df["City"].isin(["New York" , "London"])]
```

```python
# Extract people from dataset who do not live in NewYork and London
df[~df["City"].isin(["New York" , "London"])]
```

```python
#I only want poeple whose age is above 30 and their city must be paris
df[(df["City"] == "Paris") & (df["Age"] > 30 )]
```
