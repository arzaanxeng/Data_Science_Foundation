******BROADCASTING******

In NumPy, broadcasting is a powerful mechanism that allows arithmetic operations to be performed on arrays of different shapes.
Normally, to add or multiply two matrices element-by-element, they need to be the exact same size. NumPy bypasses this restriction by virtually "stretching" or duplicating the smaller array across the larger array so they have compatible shapes—all without actually copying the data in memory (which keeps it blistering fast).

**The Rules of Broadcasting**:

When operating on two arrays, NumPy compares their shapes element-wise, starting from the trailing dimensions (from right to left).
Two dimensions are compatible only if:

                                                 They are equal
                                                       or
                                            One of them is exactly 1.

If neither condition is met, NumPy throws a ValueError: operands could not be broadcast together.
The Alignment Step: 
**If one array has fewer dimensions than the other, NumPy automatically prepends dimensions of size 1 to the left side of the smaller array's shape until both shapes have the same length**


```python
# same shape
a = np.arange(6).reshape(2,3)
b = np.arange(6,12).reshape(2,3)

print(a)
print('\n')
print(b)
print('\n')
print(a+b)
```

```python
# diff shape
a = np.arange(6).reshape(2,3)
b = np.arange(3).reshape(1,3)

print(a)
print('\n')
print(b)
print('\n')
print(a+b)
```

```python
# More examples

a = np.arange(12).reshape(4,3)
b = np.arange(3)

print(a)
print('\n')
print(b)
print('\n')
print(a+b)
```

```python
a = np.arange(12).reshape(3,4)
b = np.arange(3)

print(a)
print('\n')
print(b)
print('\n')
print(a+b)
```

```python
a = np.arange(3).reshape(1,3) # (1,3) ---> (3,3)
b = np.arange(3).reshape(3,1) # (3,1) ---> (3,3)

print(a)
print('\n')
print(b)
print('\n')
print(a+b)
```
