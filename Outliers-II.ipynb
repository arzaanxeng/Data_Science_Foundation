```python
"""Let's do some outlier visualisation through boxplot"""
```

```python
Q1 = np.percentile(df1['sepal_length'] ,25)
print(Q1)
```

```python
Q3 = np.percentile(df1['sepal_length'] ,75)
print(Q3)
```

```python
IQR = Q3 - Q1
print(IQR)
```

```python
LOWER_FENCE = Q1 - (1.5*(IQR))
print(LOWER_FENCE)
```

```python
UPPER_FENCE = Q3 + (1.5*(IQR))
print(UPPER_FENCE)
```
