```python
import numpy as np
import seaborn as sns
```

```python
arr = np.array([1,2,4,5,7,9,11,13,15,17,19,21,24,25,28,30,34,36,37,220])
arr
```

```python
Q1 = np.percentile(arr,25)
print(Q1)
# 1st QUARTILE
```

```python
Q3 = np.percentile(arr,75)
# 3rd QUARTILE
```

```python
IQR = Q3 - Q1
# INTER QUARTILE

UPPER_FENCE = Q3 + (1.5*(IQR))
LOWER_FENCE = Q1 - (1.5*(IQR))
print(UPPER_FENCE)
print(LOWER_FENCE)
```

```python
arr_ = []

for i in arr:
    if i >= UPPER_FENCE or i<= LOWER_FENCE:
      continue
    else:
      arr_.append(i)
arr2 = np.array(arr_)
arr2
```

```python
sns.boxplot(arr , color = "skyblue")
# The bubble that is appearing is actually the outlier value 
```

```python
sns.boxplot(arr2 , color = "yellow")
# After 5 number summary we get the perfect boxplot
```
